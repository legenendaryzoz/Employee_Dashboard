# Advanced Analysis
The charts above compare **averages**. Averages hide spread and can show gaps that are only noise, so the cells below check data quality, add derived features, look at full distributions, and run statistical tests before drawing conclusions.

```python
from scipy import stats

# Consistent colours: one colour per entity, used in every chart
BLUE, ORANGE, GRAY = '#2a78d6', '#eb6834', '#8a8880'
GENDER_COLORS = {'Female': ORANGE, 'Male': BLUE}
sns.set_theme(style='whitegrid')
```

## 1. Data Quality Checks

```python
print('Shape:', df.shape)
print('Missing values:', df.isna().sum().sum())
print('Duplicate rows:', df.duplicated().sum())
print('Duplicate ids / emails:', df['id'].duplicated().sum(), '/', df['email'].duplicated().sum())

# Each email should contain the employee's last name
email_ok = df.apply(lambda r: r['last_name'].lower().replace("'", '').replace(' ', '') in r['email'].lower(), axis=1)
print('Emails not matching last name:', (~email_ok).sum())

# Real companies rarely start people on a weekend
weekend = df['start_date'].dt.dayofweek >= 5
print(f'Start dates on a weekend: {weekend.sum()} ({weekend.mean():.0%})')

print(f"Job titles: {df['job_title'].nunique()} unique, "
      f"{(df['job_title'].value_counts() < 5).sum()} of them have fewer than 5 employees")

df.nunique().to_frame('unique_values')
```

**Notes:** no missing values or duplicates, and emails match names. Two things to keep in mind:
- About 29% of start dates fall on a Saturday or Sunday (about 2/7, what random dates give). This suggests the dataset is **synthetic or randomly generated**, so patterns may be weak.
- Most job titles have very few employees, so per-title averages are unreliable (see section 7).

## 2. Feature Engineering

```python
# The latest start date is 2014-12-25, so 2015-01-01 is used as the snapshot date
SNAPSHOT = pd.Timestamp('2015-01-01')
df['tenure_years'] = (SNAPSHOT - df['start_date']).dt.days / 365.25
df['start_year'] = df['start_date'].dt.year

# Seniority level from the Roman numeral at the end of the title (e.g. 'Software Engineer III')
df['level'] = df['job_title'].str.extract(r'\b(IV|I{1,3})$', expand=False).fillna('No level')

df['salary_band'] = pd.cut(df['salary'], bins=[40000, 70000, 100000, 130000, 150000],
                           labels=['40-70k', '70-100k', '100-130k', '130-150k'], include_lowest=True)

df[['tenure_years', 'start_year', 'level', 'salary_band']].describe(include='all')
```

## 3. Salary Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(df['salary'], bins=22, color=BLUE, ax=axes[0])
axes[0].set_title('Salary Distribution')
axes[0].set_xlabel('Salary')
sns.boxplot(x=df['salary'], color=BLUE, ax=axes[1])
axes[1].set_title('Salary Spread')
axes[1].set_xlabel('Salary')
plt.tight_layout()
plt.show()

q1, q3 = df['salary'].quantile([0.25, 0.75])
iqr = q3 - q1
outliers = df[(df['salary'] < q1 - 1.5 * iqr) | (df['salary'] > q3 + 1.5 * iqr)]
print(f"Skewness: {df['salary'].skew():.3f} | Kurtosis: {df['salary'].kurt():.3f}")
print('IQR outliers:', len(outliers))

lo, hi = df['salary'].min(), df['salary'].max()
ks = stats.kstest(df['salary'], 'uniform', args=(lo, hi - lo))
print(f'KS test against a uniform distribution: p = {ks.pvalue:.3f} (p > 0.05 means it looks uniform)')
print(df['salary_band'].value_counts().sort_index())
```

**Notes:** salaries are spread **evenly between about 40k and 150k**: flat histogram, skewness near 0, negative kurtosis, no outliers, and a KS test that cannot reject a uniform distribution. Real salaries are usually right-skewed with a long tail at the top, which again suggests generated data.

## 4. Gender Pay Comparison

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df, x='gender', y='salary', hue='gender', palette=GENDER_COLORS, legend=False, ax=ax)
ax.set_title('Salary by Gender')
ax.set_xlabel('Gender')
ax.set_ylabel('Salary')
plt.show()

print(df.groupby('gender')['salary'].agg(['count', 'mean', 'median', 'std']).round(0))

male = df.loc[df['gender'] == 'Male', 'salary']
female = df.loc[df['gender'] == 'Female', 'salary']
t = stats.ttest_ind(male, female, equal_var=False)
u = stats.mannwhitneyu(male, female)
print(f'Welch t-test p = {t.pvalue:.3f} | Mann-Whitney U p = {u.pvalue:.3f}')

# Is either gender over-represented in some departments?
chi2_p = stats.chi2_contingency(pd.crosstab(df['department'], df['gender']))[1]
print(f'Chi-square test, department vs gender: p = {chi2_p:.3f}')
```

**Notes:** the mean gap is about 600 (Female slightly higher) and both tests give p > 0.7, so there is **no significant gender pay gap**. Gender is also spread evenly across departments (chi-square p > 0.05).

## 5. Department Pay: Which Gaps Are Real?

```python
# Compare each department with all other employees, using a Bonferroni correction for 22 tests
n_depts = df['department'].nunique()
rows = []
for dept, g in df.groupby('department'):
    rest = df.loc[df['department'] != dept, 'salary']
    p = stats.ttest_ind(g['salary'], rest, equal_var=False).pvalue
    rows.append({'department': dept, 'n': len(g), 'mean_salary': g['salary'].mean(),
                 'diff_vs_rest': g['salary'].mean() - rest.mean(), 'p_bonferroni': min(p * n_depts, 1.0)})
dept_tests = pd.DataFrame(rows).sort_values('mean_salary', ascending=False).round(3)
significant = set(dept_tests.loc[dept_tests['p_bonferroni'] < 0.05, 'department'])

# Overall tests across all departments
groups = [g['salary'] for _, g in df.groupby('department')]
anova = stats.f_oneway(*groups)
kruskal = stats.kruskal(*groups)
grand_mean = df['salary'].mean()
eta_sq = sum(len(g) * (g.mean() - grand_mean) ** 2 for g in groups) / ((df['salary'] - grand_mean) ** 2).sum()
print(f'ANOVA p = {anova.pvalue:.3f} | Kruskal-Wallis p = {kruskal.pvalue:.3f}')
print(f'Eta squared = {eta_sq:.3f} (department explains {eta_sq:.1%} of salary variation)')
print('Departments significantly different from the rest:', significant or 'none')

# Box plot sorted by median; significant departments shown in orange
order = df.groupby('department')['salary'].median().sort_values(ascending=False).index
palette = {d: ORANGE if d in significant else BLUE for d in order}
fig, ax = plt.subplots(figsize=(15, 6))
sns.boxplot(data=df, x='department', y='salary', order=order, hue='department',
            palette=palette, legend=False, ax=ax)
ax.axhline(df['salary'].median(), color=GRAY, ls='--', lw=1, label='Company median')
ax.set_title('Salary by Department (orange = significantly different after Bonferroni correction)')
ax.set_xlabel('Department')
ax.set_ylabel('Salary')
ax.tick_params(axis='x', rotation=45)
ax.legend(loc='upper right', framealpha=0.95)
plt.tight_layout()
plt.show()

dept_tests
```

**Notes:** department differences are only borderline significant overall (p about 0.04) and explain about **3% of salary variation**. The one gap that holds up after correcting for 22 comparisons is **Outdoors**, which pays about 15k more than the rest of the company. The earlier bar chart of averages made the department gaps look bigger than they are.

## 6. Region, Tenure and Seniority

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
df['region_id'].value_counts().sort_index().plot(kind='bar', color=BLUE, ax=axes[0])
axes[0].set_title('Employees per Region')
axes[0].set_xlabel('Region')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=0)
sns.boxplot(data=df, x='region_id', y='salary', color=BLUE, ax=axes[1])
axes[1].set_title('Salary by Region')
axes[1].set_xlabel('Region')
axes[1].set_ylabel('Salary')
plt.tight_layout()
plt.show()

region_p = stats.f_oneway(*[g['salary'] for _, g in df.groupby('region_id')]).pvalue
dept_region_p = stats.chi2_contingency(pd.crosstab(df['department'], df['region_id']))[1]
print(f'ANOVA salary by region: p = {region_p:.3f}')
print(f'Chi-square department vs region: p = {dept_region_p:.3f}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.regplot(data=df, x='tenure_years', y='salary', ax=axes[0],
            scatter_kws={'alpha': 0.35, 's': 15, 'color': BLUE}, line_kws={'color': ORANGE, 'lw': 2})
axes[0].set_title('Salary vs Tenure')
axes[0].set_xlabel('Tenure (years)')
axes[0].set_ylabel('Salary')

df['start_year'].value_counts().sort_index().plot(kind='bar', color=BLUE, ax=axes[1])
axes[1].set_title('Hires per Year')
axes[1].set_xlabel('Start year')
axes[1].set_ylabel('Hires')
plt.tight_layout()
plt.show()

r, p_r = stats.pearsonr(df['tenure_years'], df['salary'])
rho, p_rho = stats.spearmanr(df['tenure_years'], df['salary'])
print(f'Pearson r = {r:.3f} (p = {p_r:.3f}) | Spearman rho = {rho:.3f} (p = {p_rho:.3f})')
```

```python
level_order = ['I', 'II', 'III', 'IV', 'No level']
fig, ax = plt.subplots(figsize=(9, 5))
sns.boxplot(data=df, x='level', y='salary', order=level_order, color=BLUE, ax=ax)
ax.set_title('Salary by Seniority Level (from job title)')
ax.set_xlabel('Level')
ax.set_ylabel('Salary')
plt.show()

leveled = df[df['level'] != 'No level']
level_p = stats.kruskal(*[g['salary'] for _, g in leveled.groupby('level')]).pvalue
print(df.groupby('level')['salary'].agg(['count', 'mean', 'median']).reindex(level_order).round(0))
print(f'Kruskal-Wallis across levels I-IV: p = {level_p:.3f}')
```

**Notes:** region, tenure and seniority level have **no measurable effect on salary**. Tenure correlation is about 0 and Level IV earns no more than Level I (Kruskal-Wallis p = 0.85). In a real company, seniority and tenure are usually the strongest salary drivers, so these results point to randomly generated salaries rather than real pay policy.

## 7. Job Titles With Enough Data
The earlier chart ranked all job titles by average salary, but most titles have only a few employees, so that ranking is mostly noise. Here only titles with at least 10 employees are shown.

```python
MIN_N = 10
title_counts = df['job_title'].value_counts()
big_titles = title_counts[title_counts >= MIN_N].index
title_order = df[df['job_title'].isin(big_titles)].groupby('job_title')['salary'].median().sort_values(ascending=False).index

fig, ax = plt.subplots(figsize=(10, max(5, len(title_order) * 0.45)))
sns.boxplot(data=df[df['job_title'].isin(big_titles)], y='job_title', x='salary',
            order=title_order, color=BLUE, ax=ax)
ax.set_yticks(range(len(title_order)))
ax.set_yticklabels([f'{t} (n={title_counts[t]})' for t in title_order])
ax.set_title(f'Salary by Job Title (titles with at least {MIN_N} employees)')
ax.set_xlabel('Salary')
ax.set_ylabel('')
plt.tight_layout()
plt.show()

title_p = stats.kruskal(*[g['salary'] for _, g in df[df['job_title'].isin(big_titles)].groupby('job_title')]).pvalue
print(f'Kruskal-Wallis across these titles: p = {title_p:.3f}')
```

| 2 | Salaries are evenly spread between about 40k and 150k (median about 96k), with no outliers | Flat histogram, skew about 0, KS p about 0.10 |

**Caveat:** uniform salaries, about 29% weekend start dates, and no link between seniority and pay strongly suggest this is a **synthetic dataset**. The methods here suit real HR data, but these results should not be used for real compensation decisions.

**Possible next steps:** with real data, model salary on department, level, tenure and region using regression (`statsmodels`), and track the gender gap within the same job title and level rather than overall.
